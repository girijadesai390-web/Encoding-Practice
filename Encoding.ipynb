```python
import pandas as pd
```

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
```

```python
data = {
    'Size': ['Medium', 'Small', 'Large', 'Medium', 'Small'],  # Ordinal feature
    'Color': ['Red', 'Blue', 'Green', 'Blue', 'Red']           # Nominal feature
}
df = pd.DataFrame(data)
```

```python
df
```

```python
label_encoder = LabelEncoder()
df['Size_Encoded'] = label_encoder.fit_transform(df['Size'])
```

```python
df
```

```python
df['Color_Encoded'] = label_encoder.fit_transform(df['Color'])
```

```python
one_hot_encoder = OneHotEncoder(sparse_output=False)
Size_encoded = one_hot_encoder.fit_transform(df[['Size']])
```

```python
df
```

```python
Size_encoded
```

```python
encoded_columns = one_hot_encoder.get_feature_names_out(['Size'])
ohe_df = pd.DataFrame(Size_encoded, columns=encoded_columns)
```

```python
ohe_df
```

```python
color_encoded = one_hot_encoder.fit_transform(df[['Color']])
encoded_columns = one_hot_encoder.get_feature_names_out(['Color'])
ohe_df2 = pd.DataFrame(color_encoded, columns=encoded_columns)
new_df= pd.concat([ohe_df,ohe_df2],axis=1)
new_df
```

```python
final_df= pd.concat([df,new_df],axis=1)
final_df
```
